# RAG4 - Sistema de Notificaciones Email
## Vulcanizadora con SMTP Gmail
---
Envía reportes de métricas y resumen de consultas por correo automáticamente.

**Ejecutar todas las celdas en orden.**

```python
#CELDA 2 
import smtplib
import os
from email.mime.multipart import MIMEMultipart
from email.mime.text import MIMEText
from datetime import datetime
from dotenv import load_dotenv

load_dotenv()

GMAIL_USER     = os.getenv("GMAIL_USER")
GMAIL_PASSWORD = os.getenv("GMAIL_APP_PASSWORD")
EMAIL_DESTINO  = os.getenv("EMAIL_DESTINO")

print(f"Gmail configurado: {'✅' if GMAIL_USER else '❌'}")
print(f"Contraseña cargada: {'✅' if GMAIL_PASSWORD else '❌'}")
print(f"Destino: {EMAIL_DESTINO if EMAIL_DESTINO else '❌'}")
```

```python
#CELDA 3
class EnviadorEmail:
    """Envía reportes del sistema de vulcanización por Gmail SMTP."""

    def __init__(self):
        self.gmail_user     = GMAIL_USER
        self.gmail_password = GMAIL_PASSWORD
        self.destino        = EMAIL_DESTINO

    def enviar(self, asunto: str, cuerpo_html: str) -> bool:
        try:
            msg = MIMEMultipart("alternative")
            msg["Subject"] = asunto
            msg["From"]    = self.gmail_user
            msg["To"]      = self.destino
            msg.attach(MIMEText(cuerpo_html, "html"))

            with smtplib.SMTP_SSL("smtp.gmail.com", 465) as servidor:
                servidor.login(self.gmail_user, self.gmail_password)
                servidor.sendmail(self.gmail_user, self.destino, msg.as_string())

            print(f"✅ Email enviado a {self.destino}")
            return True

        except Exception as e:
            print(f"❌ Error al enviar: {e}")
            return False

print("Clase EnviadorEmail lista ✅")
```

```python
#CELDA 4
def generar_reporte_html(consultas: list, metricas: dict) -> str:
    fecha = datetime.now().strftime("%d/%m/%Y %H:%M")

    filas_consultas = ""
    for item in consultas:
        filas_consultas += f"""
        <tr>
            <td style="padding:8px;border-bottom:1px solid #eee">{item['consulta']}</td>
            <td style="padding:8px;border-bottom:1px solid #eee;text-align:center">{item['tipo']}</td>
            <td style="padding:8px;border-bottom:1px solid #eee;text-align:center">
                {'✅' if item['exitoso'] else '❌'}
            </td>
        </tr>"""

    html = f"""
    <html><body style="font-family:Arial,sans-serif;max-width:600px;margin:auto">
        <div style="background:#1a1a2e;padding:20px;border-radius:8px 8px 0 0">
            <h1 style="color:#fff;margin:0">🔧 Vulcanizadora - Reporte IA</h1>
            <p style="color:#aaa;margin:5px 0 0">{fecha}</p>
        </div>
        <div style="background:#f8f9fa;padding:20px">
            <h2 style="color:#333">📊 Métricas del Sistema</h2>
            <table width="100%" style="border-collapse:collapse">
                <tr>
                    <td style="padding:10px;background:#fff;border-radius:8px;text-align:center;width:25%">
                        <div style="font-size:28px;font-weight:bold;color:#1a1a2e">{metricas.get('total_peticiones', 0)}</div>
                        <div style="color:#666;font-size:12px">Total consultas</div>
                    </td>
                    <td style="padding:10px;background:#fff;border-radius:8px;text-align:center;width:25%">
                        <div style="font-size:28px;font-weight:bold;color:#27ae60">{metricas.get('tiempo_promedio_ms', 0)}ms</div>
                        <div style="color:#666;font-size:12px">Tiempo promedio</div>
                    </td>
                    <td style="padding:10px;background:#fff;border-radius:8px;text-align:center;width:25%">
                        <div style="font-size:28px;font-weight:bold;color:#e74c3c">{metricas.get('tasa_errores_pct', 0)}%</div>
                        <div style="color:#666;font-size:12px">Tasa de errores</div>
                    </td>
                    <td style="padding:10px;background:#fff;border-radius:8px;text-align:center;width:25%">
                        <div style="font-size:28px;font-weight:bold;color:#f39c12">{metricas.get('total_tokens', 0)}</div>
                        <div style="color:#666;font-size:12px">Tokens usados</div>
                    </td>
                </tr>
            </table>
            <h2 style="color:#333;margin-top:20px">💬 Resumen de Consultas</h2>
            <table width="100%" style="border-collapse:collapse;background:#fff;border-radius:8px">
                <tr style="background:#1a1a2e;color:#fff">
                    <th style="padding:10px;text-align:left">Consulta</th>
                    <th style="padding:10px">Tipo</th>
                    <th style="padding:10px">Estado</th>
                </tr>
                {filas_consultas}
            </table>
        </div>
        <div style="background:#1a1a2e;padding:15px;border-radius:0 0 8px 8px;text-align:center">
            <p style="color:#aaa;margin:0;font-size:12px">Sistema de Observabilidad - Vulcanizadora IA</p>
        </div>
    </body></html>
    """
    return html

print("Generador HTML listo ✅")
```

```python
#CELDA 5
consultas_prueba = [
    {"consulta": "¿Cuánto cuesta un parche interno?",   "tipo": "parche",           "exitoso": True},
    {"consulta": "Necesito cambiar los 4 neumáticos",   "tipo": "cambio_neumatico", "exitoso": True},
    {"consulta": "¿Hacen alineación computarizada?",    "tipo": "alineacion",       "exitoso": True},
    {"consulta": "¿Cuál es el horario de atención?",    "tipo": "consulta_horario", "exitoso": True},
    {"consulta": "¿Cuánto cuesta la revisión técnica?", "tipo": "consulta_general", "exitoso": False},
]

metricas_prueba = {
    "total_peticiones":   5,
    "tiempo_promedio_ms": 461.4,
    "tiempo_maximo_ms":   820.0,
    "tiempo_minimo_ms":   210.0,
    "total_tokens":       1250,
    "tasa_errores_pct":   20.0,
}

email = EnviadorEmail()
html  = generar_reporte_html(consultas_prueba, metricas_prueba)

email.enviar(
    asunto="🔧 Reporte Vulcanizadora IA - " + datetime.now().strftime("%d/%m/%Y"),
    cuerpo_html=html
)
```
